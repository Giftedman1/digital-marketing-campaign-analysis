```python
import pandas as pd
import numpy as np
import os
```

```python
# Load all CSV files from the data folder
campaigns = pd.read_csv('data/campaigns.csv')
customers = pd.read_csv('data/customers.csv')
purchases = pd.read_csv('data/purchases.csv')
sessions = pd.read_csv('data/website_sessions.csv')
daily = pd.read_csv('data/daily_performance.csv')
ab_test = pd.read_csv('data/ab_test.csv')
platforms = pd.read_csv('data/platforms.csv')
campaign_types = pd.read_csv('data/campaign_types.csv')
industries = pd.read_csv('data/industries.csv')


# Store in a dictionary for easy access
datasets = {
    'Campaigns': campaigns,
    'Customers': customers,
    'Purchases': purchases,
    'Sessions': sessions,
    'Daily Performance': daily,
    'A/B Test': ab_test,
    'Platforms': platforms,
    'Campaign Types': campaign_types,
    'Industries': industries
}
```

```python
# Check row counts for each dataset
print("Row counts:")]
print("-" * 30)

for name, df in datasets.items():
    rows = len(df)
    print(f"{name}: {rows:,} rows")
```

```python
# Check if any dataset has missing values
print("Missing values:")
print("-" * 30)

has_missing = False

for name, df in datasets.items():
    missing = df.isnull().sum().sum()
    if missing > 0:
        print(f"{name}: {missing} missing values")
        has_missing = True
    else:
        print(f"{name}: No missing values")

if not has_missing:
    print("All datasets are complete. No missing values found.")
```

```python
# Check if any dataset has duplicate rows
print("Duplicate rows:")
print("-" * 30)

has_duplicates = False

for name, df in datasets.items():
    duplicates = df.duplicated().sum()
    if duplicates > 0:
        print(f"{name}: {duplicates} duplicate rows")
        has_duplicates = True
    else:
        print(f"{name}: No duplicate rows")

if not has_duplicates:
    print("No duplicate rows found.")
```

```python
# Check if primary keys are unique
print("Primary keys:")
print("-" * 30)

# Define which column is the primary key for each table
primary_keys = {
    'Campaigns': 'CampaignID',
    'Customers': 'CustomerID',
    'Purchases': 'PurchaseID',
    'Sessions': 'SessionID',
    'A/B Test': 'TestID'
}

for name, pk in primary_keys.items():
    df = datasets[name]
    duplicates = df[pk].duplicated().sum()
    if duplicates > 0:
        print(f"{name}: {duplicates} duplicate {pk}")
    else:
        print(f"{name}: Unique {pk}")
```

```python
# Check if dates are valid
print("Date validation:")
print("-" * 30)

# Convert to datetime
campaigns['StartDate'] = pd.to_datetime(campaigns['StartDate'])
campaigns['EndDate'] = pd.to_datetime(campaigns['EndDate'])

# Check if end date is after start date
dates_valid = (campaigns['EndDate'] >= campaigns['StartDate']).all()
print(f"End date >= Start date: {dates_valid}")

# Show date range
start_min = campaigns['StartDate'].min()
start_max = campaigns['StartDate'].max()
end_min = campaigns['EndDate'].min()
end_max = campaigns['EndDate'].max()

print(f"Start date range: {start_min.date()} to {start_max.date()}")
print(f"End date range: {end_min.date()} to {end_max.date()}")
```
